**Class balance:** 164 patients without disease (54%) vs 139 with disease (46%). The classes are close to balanced, so accuracy is meaningful, but we still use stratified splits and track precision/recall/ROC-AUC. We also try `class_weight='balanced'` during tuning.

```python
display(Image(filename=str(config.FIGURES_DIR / '03_feature_histograms.png'), width=900))
```

**Histograms:** patients with disease are older, reach a lower maximum heart rate (`thalach`) and show more exercise-induced ST depression (`oldpeak`); `ca` is strongly shifted. Cholesterol and resting blood pressure overlap heavily — weak predictors on their own. `oldpeak` and `chol` are right-skewed, so features are standardised (and tree models are insensitive to skew).

```python
display(Image(filename=str(config.FIGURES_DIR / '04_correlation_heatmap.png'), width=760))
display(Image(filename=str(config.FIGURES_DIR / '05_target_correlation.png'), width=560))
```

**Correlations:** strongest linear associations with the target are `thal` (0.53), `ca` (0.46), `exang` (0.43), `oldpeak` (0.42), `thalach` (−0.42) and `cp` (0.41). No pair of predictors is highly collinear (|r| < 0.6), so no feature needs to be dropped. Note that `cp`, `thal`, `restecg` and `slope` are nominal codes — their Pearson correlation is only indicative, which is why they are one-hot encoded.

```python
display(Image(filename=str(config.FIGURES_DIR / '06_categorical_disease_rate.png'), width=900))
```

**Categorical features:** asymptomatic chest pain (`cp=4`) has a ~73% disease rate versus ~20–30% for the other types — the effect is non-monotonic in the code, confirming one-hot encoding. Reversible thallium defect (`thal=7`) and a flat/down-sloping ST segment are strong signals. `fbs` barely changes the rate. `restecg=1` has only 4 patients, so its estimate is noisy.

```python
display(Image(filename=str(config.FIGURES_DIR / '07_boxplots.png'), width=900))
display(Image(filename=str(config.FIGURES_DIR / '08_age_vs_max_hr.png'), width=600))
```

**Outliers & feature idea:** a few high cholesterol (564 mg/dl) and `oldpeak` values exist but are clinically plausible, so they are kept (scaling + regularisation handle them). The age-vs-heart-rate plot shows that diseased patients fall further below the age-predicted maximum (220 − age). This motivates the engineered feature `hr_achieved_ratio = thalach / (220 − age)` and the rate-pressure index `thalach × trestbps / 1000`; whether they help is decided by the hyperparameter search (`features__enabled`).

## 4. Take-aways for modelling
* 13 inputs: 6 continuous (scaled), 4 nominal (one-hot), 3 binary (passed through).
* Missing `ca`/`thal` → median / most-frequent imputation fitted inside CV folds.
* Near-balanced target → stratified 80/20 split + stratified 5-fold CV, ROC-AUC as the selection metric, recall tracked because a missed diagnosis is costlier than a false alarm.
