## 2. Baseline comparison with default hyperparameters (5-fold stratified CV)

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.dummy import DummyClassifier
from src.evaluate import CV_SCORING
cv = StratifiedKFold(5, shuffle=True, random_state=config.RANDOM_STATE)
rows = {}
for name, est in {'majority': DummyClassifier(), 'logistic_regression': LogisticRegression(max_iter=2000, solver='liblinear'),
                  'random_forest': RandomForestClassifier(random_state=42), 'gradient_boosting': GradientBoostingClassifier(random_state=42)}.items():
    res = cross_validate(build_pipeline(est), X_train, y_train, cv=cv, scoring=CV_SCORING)
    rows[name] = {m: res[f'test_{m}'].mean() for m in CV_SCORING}
pd.DataFrame(rows).T.round(3)
```

Untuned, logistic regression is already competitive with the ensembles — expected for a small (242 training rows), mostly linearly separable tabular dataset. Random forest overfits slightly with unlimited depth. This motivates tuning regularisation for LR and depth / leaf size for the tree models.

## 3. Tuning strategy (implemented in `src/train.py`)
| Model | Search | Space | Candidates |
|---|---|---|---|
| Logistic Regression | Grid | C ∈ {0.01…10}, penalty L1/L2, class_weight, engineered features on/off | 56 |
| Random Forest | Randomised | n_estimators, max_depth, min_samples_leaf, max_features, class_weight, engineered on/off | 30 |
| Gradient Boosting | Randomised | n_estimators, learning_rate, max_depth, subsample, min_samples_leaf, engineered on/off | 25 |

Each candidate is scored by stratified 5-fold CV on the training split for accuracy, precision, recall, F1 and ROC-AUC; the refit metric is ROC-AUC (threshold-independent). The champion family is chosen by **mean CV ROC-AUC** — not test score — so the 61-patient hold-out set stays an unbiased final check.

## 4. Results logged by the training run (read back from MLflow)

```python
import mlflow
mlflow.set_tracking_uri(config.MLFLOW_TRACKING_URI)
runs = mlflow.search_runs(experiment_names=[config.MLFLOW_EXPERIMENT], filter_string="tags.stage = 'tuning'")
cols = ['tags.mlflow.runName','metrics.cv_roc_auc_mean','metrics.cv_roc_auc_std','metrics.cv_accuracy_mean','metrics.cv_recall_mean','metrics.test_accuracy','metrics.test_precision','metrics.test_recall','metrics.test_roc_auc']
runs[cols].sort_values('metrics.cv_roc_auc_mean', ascending=False).round(3)
```

```python
trials = mlflow.search_runs(experiment_names=[config.MLFLOW_EXPERIMENT], filter_string="tags.stage = 'trial'")
print(len(trials), 'tuning trials logged as nested runs')
trials['family'] = trials['tags.mlflow.runName'].str.replace(r'-trial-\d+', '', regex=True)
trials.groupby(['family', 'params.engineered_features'])['metrics.cv_roc_auc_mean'].agg(['mean','max','count']).round(3)
```

The ablation built into the search shows the engineered features give no consistent gain for LR/RF (the information is already in `thalach`, `age` and `trestbps`) but are selected by gradient boosting. Keeping the switch as a hyperparameter lets the data decide rather than hard-coding it.

## 5. Model selection decision
* **Champion: regularised logistic regression** (L2, C = 0.3, balanced class weights). Highest mean CV ROC-AUC (0.907 ± 0.019) with the lowest variance across folds; on the hold-out set ROC-AUC 0.961, recall 0.93.
* Random forest is within noise on ROC-AUC (0.905) and slightly better on hold-out accuracy, but is less interpretable, larger and slower — so with equal performance the simpler model wins.
* The coefficients are clinically sensible: more major vessels on fluoroscopy, male sex, asymptomatic chest pain, a reversible thallium defect and exercise angina push risk up; a normal thallium scan, an up-sloping ST segment and a higher max heart rate push it down.
